Dependent language ka matlab yahan high-level programming languages like (C, C++) se hai jo directly machine code nahi hoti, balkay compiler ke through convert hoti hain.

```python
#include <stdio.h>
int main(){
    printf("Hello World");
    return 0;
}
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- Ab compiler kya karta hai?
- Ye code har system par same nahi chalta directly, balkay compiler isko convert karta hai:

    - Windows (x86 CPU) 👉 Machine Code A <p>10110000 00000101<br>00000001 11000010<br>

   - Linux (x86 CPU) 👉 Machine Code B <p>11001000 00010101<br>00000001 10101010</p>
     
    - ARM (Mobile) 👉 Machine Code C <p>11100011 00010001<br>01010100 11000001</p>
 
- 👉 Same C code but 3 different machine codes ;

<h1 style="text-align:center; color:#005bbd; font-size:20px; font-family:Sans-serif; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px Black;">
   Byte Code
</h1>

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

Bytecode ek “beech ka code” (intermediate code) hota hai jo human code (Python/Java) aur machine code ke darmiyan hota hai.

- Ye directly CPU samajh nahi sakta
- Isko chalane ke liye VM (Virtual Machine) chahiye

<body style="font-family: Sans-serif;">

Java Code

```python
public class HelloWorld {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
```

```python
%%bash 
echo '''public class HelloWorld {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
''' > HelloWorld.java
```

```python
!ls HelloWorld.java
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

Compile Java Code to Byte Code

```python
!javac HelloWorld.java
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

JVM (Java Virtual Machine) ek software engine hai jo Java bytecode ko run karta hai.

👉 Simple words mein:
- JVM ek virtual computer hai jo Java programs chalata hai.

<img src="./images/2.jpg" style="display: block;margin-left: auto; margin-right: auto;width:90%">

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

JVM ka main kaam:

✔ 1. Bytecode run karna
  - .class file ko read karta hai
  - usko execute karta hai
  
✔ 2. Platform independent banana
- Windows, Linux, Mac sab pe same bytecode chalta hai
- Because ? JVM har OS ke liye alag hota hai

✔ 3. Memory manage karna
- JVM khud handle karta hai:
    - Heap (objects)
    - Stack (functions)
    - Garbage Collection (unused memory delete)

✔ 4. Security provide karna
- Bytecode verify karta hai
- unsafe code ko reject karta hai
