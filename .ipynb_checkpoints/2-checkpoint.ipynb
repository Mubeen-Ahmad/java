- JVM
    - JVM (Java Virtual Machine) Java ka core engine hai jo program ko run karta hai. 
    - JVM woh system hai jo .class bytecode ko le kar usay OS par execute karta hai.

```python
%%bash 
echo '''public class HelloWorld {
    public static void main(String[] args) {
        System.out.println("Hello World");
    }
}
''' > HelloWorld.java
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 16px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- .java File
    - Ye source code file hoti hai jo programmer likhta hai.

```python
!ls HelloWorld.java
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 16px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- javac Java Compiler hai.
    - Ye .java file ko compile karke .class file banata hai.

```python
!javac HelloWorld.java
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 16px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">
        
- .class file mein Bytecode hota hai.


```python
!ls HelloWorld.class
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 16px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">
        
- .class file mein Bytecode hota hai.
    - `javap -c`  bytecode ko human-readable form mein dikhata hai.


```python
!javap -c HelloWorld
```

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- JVM Security Sandbox
    - JVM program ko ek controlled box (sandbox) mein run karta hai.
    - Means:
        - Program direct OS ko access nahi kar sakta
        - File system, memory, network access limited hota hai
        - Sirf allowed operations hoti hain
        - Illegal memory access detect karta hai
        - corrupted code reject karta hai
        ---
    - Garbage Collector (Memory Management)
        - Jo objects use nahi ho rahe unko remove karta hai
        - Memory free karta hai automatically

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- Platform Independence Enable karna
    - JVM har OS ke liye alag hota hai:
    - Windows JVM
    - Linux JVM
    - Mac JVM
    - etc


- Is liye same bytecode har OS par chal jata hai

<h1 style="text-align:center; color:#005bbd; font-size:20px; font-family:Sans-serif; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px Black;">
   Hybrid Model Java
</h1>

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

Java ko aik Hybrid Model kaha jata hai kyunke yeh Compiled aur Interpreted dono tarah se kaam karti hai. 

1.  Source Code to Bytecode (Compiler)
    - Jab Java code likhte hain, to sab se pehle aik compiler use hota hai (jisay javac kehte hain).
    - Yeh compiler aapke source code ko Bytecode (.class file) mein convert kar deta hai.
    - Yeh Bytecode "platform independent" hota hai, yani yeh kisi bhi OS par chal sakta hai.
    ---

2. Bytecode se Machine Code (Hybrid Execution)
    - hybrid model JVM (Java Virtual Machine) ke andar hota hai.
    - Old Days mein JVM sirf aik Interpreter use karta tha jo code ko line-by-line machine code mein badalta tha, jo ke slow hota tha.
   ---
    -  Modern Java mein speed barhane ke liye JVM 2 cheezon ka combination use karta hai:
        -  Interpreter:
            -   Yeh code ko line-by-line convert aur run karta rehta hai taake program foran start ho jaye. <br> <br>

        - JIT (Just-In-Time) Compiler:
            - Yeh aik bahut smart aur optimized compiler hai.
            - JVM dekhta hai ke code ka kaun sa hissa bar-bar use ho raha hai (frequent code).
            - JIT us frequent code ko aik hi baar mein machine code mein compile kar deta hai taake agli baar use line-by-line interpret na karna paray aur program fast chale.

<h1 style="text-align:center; color:#005bbd; font-size:20px; font-family:Sans-serif; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px Black;">
   JRE
</h1>

<body style="font-family: Sans-serif;">
    <div style="color: black; font-size: 18px; font-style: oblique; text-shadow: 0 0 3px white, 0 0 1px black; padding: 20px;">

- JRE (Java Runtime Environment) ek runtime environment hai jo Java applications ko run karne ke liye zaroori components provide karta hai.
  

- JRE ke 2 main kaam/parts hote hain:

- 1. JVM (Java Virtual Machine)
      - Java bytecode run karta hai
      - Program execute karta hai
      - Memory manage karta hai (GC etc.)

- 2 Java Libraries (Runtime Libraries)
    - Ready-made code provide karti hain
    - Example:
        - System.out.println
        - ArrayList
        - String
        - Thread
